# Data study

What are dataset sizes ? Item size ? Variance ?

```python
import pandas as pd
```

```python
def get_dataset_stats(series_tokens, series_atoms):
    """Calculates metrics for tokens and atoms and returns them as a Pandas Series."""
    return pd.Series(
        {
            "Dataset size": len(series_tokens),
            "Max/Min tokens": (series_tokens.max(), series_tokens.min()),
            "Max/Min atoms": (series_atoms.max(), series_atoms.min()),
            "Mean tokens": series_tokens.mean(),
            "Mean atoms": series_atoms.mean(),
            "Std tokens": series_tokens.std(),
            "Std atoms": series_atoms.std(),
        }
    )

def latex_code(df, dataset: str):
    return df.to_latex(
        index=True,  # Keeps the metric names as row labels
        column_format="l" + "c" * len(df.columns),  # Left align metrics, center data
        caption=f"{dataset} Analysis: Complexes lengths",
        label=f"tab:{dataset}_analysis",
        position="htbp",
        escape=False,  # Allows the ± symbol to render cleanly
    )
```

## Skempi

```python
skempi_df = pd.read_csv(
    "../lengths/skempi_0/train_lengths_with_atoms.csv",
    header=None,
    names=["tokens", "atoms"],
)

skempi_notrunc_df = pd.read_csv(
    "../lengths/skempi_0_no_truncation/train_lengths_with_atoms.csv",
    header=None,
    names=["tokens", "atoms"],
)

stats_skempi = get_dataset_stats(skempi_df["tokens"], skempi_df["atoms"])
stats_skempi_no_trunc = get_dataset_stats(skempi_notrunc_df["tokens"], skempi_notrunc_df["atoms"])

# 3. Combine into a single DataFrame where Datasets are Columns and Metrics are Rows
summary_df = pd.DataFrame(
    {
        "SKEMPI (Truncated)": stats_skempi,
        "SKEMPI (not Truncated)": stats_skempi_no_trunc,
    }
)

print(summary_df)
print(latex_code(summary_df), "skempi")
```
