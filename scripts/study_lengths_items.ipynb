# Data study

What are dataset sizes ? Item size ? Variance ?

```python
import pandas as pd
```

```python
def dict_analyse(df):
    analyse = {}
    analyse["dataset_size"] = len(df)

    analyse["max_min_token"] = (df[0].max(),df[0].min())
    analyse["max_min_atom"] = (df[1].max(),df[1].min())

    analyse["mean_token"] = df[0].mean()
    analyse["mean_atom"] = df[1].mean()

    analyse["ecart type tokens"] = df[0].std()
    analyse["ecart type atoms"] = df[1].std()

    return analyse

def display(analyse):
    for key,value in analyse.items():
        print(key, " : ", value)
```

## Skempi

```python
df = pd.read_csv("../lengths/skempi_0/train_lengths_with_atoms.csv", header=None) # token, atom counts
print(df.head())
display(dict_analyse(df))
```

## LBA

```python
df = pd.read_csv("../lengths/lba_identity_30/train_lengths_with_atoms.csv", header=None) # token, atom counts
print(df.head())
display(dict_analyse(df))
```

## PDB

```python
df = pd.read_csv("../lengths/pdb/train_lengths_with_atoms.csv", header=None) # token, atom counts
print(df.head())
display(dict_analyse(df))
```
